# Communication with the APB bus
### Requirements:
* PySerial (installation:
    * activate environment
    * pip install pyserial
* Access rights for the serial port

```python
import serial
import glob
import time
```

```python
# List serial ports
glob.glob('/dev/tty.*')
```

```python
# Open the serial port for the UART (change the port name accordingly)
ser = serial.Serial('/dev/tty.usbserial-21200')
```

```python
# Configure the serial port
ser.baudrate=115200
ser.stopbits=2
ser.timeout=0.5
```

```python
# Read until the serial port buffer is empty
def empty():
    try:
        while(ser.read() != b''):
            pass
    except:
        pass
```

```python
# Read from an address on the APB bus
def read_from_addr(addr):
    request = ("r " + hex(addr) + "\n").encode("utf-8")
    ser.write(request)
    ser.flush()
    time.sleep(0.01)
    ser.read_until()
    response = ser.read_until()
    response_str = response.decode("utf-8")
    value = int(response_str.split("data:")[1].strip(), base=16)
    return(value)
```

```python
# Write to an address on the APB bus
def write_to_addr(addr, data):
    request = ("w " + hex(addr) + " " + hex(data) + "\n").encode("utf-8")
    ser.write(request)
    ser.flush()
    time.sleep(0.01)
    ser.read_until()
    ser.read_until()
```

```python
# At the very beginning:  empty the buffer
empty()
```

```python
# for reading (returns the result as an int)
read_from_addr(0xf000000C)
```

```python
# for writing
write_to_addr(0xf0000004, 123)
```

```python
# at the very end of the program: close the port.
ser.close()
```
